## Date preprocessing of neighborhood parks and reserves

```python
import geopandas as gpd

# Load the shapefile

shape_data_parks = gpd.read_file('../data/landing/external_data/shapefile/Parks_reserves/ll_gda94/esrishape/whole_of_dataset/victoria/CROWNLAND/PARKRES.shp')

# Display rows of the data
print(shape_data_parks)
```

## Data preprocessing

```python
# Detect missing values
print(shape_data_parks.isnull().sum())

# Drop or fill missing values
shape_data_parks.dropna(inplace=True)
shape_data_parks
```

```python
#rename the column of 'NAME'
shape_data_parks.rename(columns={'NAME': 'park_reserve_name'}, inplace=True)
shape_data_parks = shape_data_parks[['geometry', 'park_reserve_name']]
# Get the centroids for each location
shape_data_parks['parkreserve_centroid'] = shape_data_parks['geometry'].apply(lambda x: (x.centroid.y, x.centroid.x))
shape_data_parks
```

```python
from shapely import wkt

# Assuming the equivalent of 'station_centroid' for parks is 'geometry'
shape_data_parks['parkreserve_centroid'] = shape_data_parks['geometry'].centroid

# Extract the latitude and longitude from the centroids
shape_data_parks['parkreserve_coordinates'] = shape_data_parks['parkreserve_centroid'].apply(lambda geom: [geom.y, geom.x])

# Format the coordinates into the desired string format
shape_data_parks['parkreserve_coordinates'] = shape_data_parks['parkreserve_coordinates'].apply(lambda coords: f"[{coords[0]}, {coords[1]}]")
shape_data_parks
```

```python
# save as the CSV file into external_data directory for further use
shape_data_parks.to_csv("../data/raw/external_data/parks_data.csv", index=False)
# Computing counts of parks and reserves
parks_counts = shape_data_parks['park_reserve_name'].value_counts()
print(parks_counts)
```

```python
# Keep a unique record of each park name
shape_data_parks= shape_data_parks.drop_duplicates(subset=['park_reserve_name'])  

# Extract the desired columns
shape_data_parks = shape_data_parks[['park_reserve_name', 'parkreserve_coordinates']]

# Save the updated data to a CSV file into raw folder for further usage
shape_data_parks.to_csv("../data/raw/external_data/parks_data.csv", index=False)
shape_data_parks
```

## Find the closest park or reserve

```python
import pandas as pd
# read property data and parks data
property_data = pd.read_csv("../data/curated/merged_data/merged_data_with_facility.csv")
parks_data = pd.read_csv("../data/raw/external_data/parks_data.csv")
parks_data = parks_data[['park_reserve_name', 'parkreserve_coordinates']]
parks_data
property_data
```

```python
from geopy.distance import great_circle
import numpy as np

# Convert a string formatted as "[latitude, longitude]" into a tuple (latitude, longitude)
def parse_coordinates(coord_str):
    parts = coord_str.strip('[]').split(',')
    return float(parts[0]), float(parts[1])

# Convert string representations of coordinates into actual coordinate tuples
property_coordinates = property_data['coordinates'].apply(parse_coordinates)
park_coordinates = parks_data['parkreserve_coordinates'].apply(parse_coordinates)

# For a given property coordinate, determine the single nearest park and its distance
def find_nearest_park(property_coord):
    if np.isnan(property_coord).any():
        return None, float('inf')  # If invalid property coordinates, return None and infinite distance
    
    distances = [
        great_circle(property_coord, park_coord).kilometers
        if not np.isnan(park_coord).any() else float('inf')
        for park_coord in park_coordinates
    ]
    
    # Identify the index of the closest park
    closest_index = np.argmin(distances)
    
    # Extract the closest park's coordinates
    closest_park = parks_data.loc[closest_index, 'parkreserve_coordinates']
    
    return closest_park, distances[closest_index]

# Compute closest park and its distance for each property
property_data['closest_park'], property_data['park_distance(KM)'] = zip(*property_coordinates.apply(find_nearest_park))

# Display the modified property_data dataframe
property_data
```

```python
#save the updated file into curated folder 
property_data.to_csv('../data/curated/merged_data/merged_data_with_facility.csv', index=False)
```
