create folders under `../data/`

```python
import os
```

```python
DATA_STEP = ['landing', 'raw', 'curated']
DATA_SOURCE = ['domain_data', 'external_data']

for data_step in DATA_STEP:
    for data_source in DATA_SOURCE:
        directory = f'../../real-estate-industry-project-open-source-industry-project-22/data/{data_step}/{data_source}'
        gitkeep_path = os.path.join(directory, '.gitkeep')

        if not os.path.exists(directory):
            os.makedirs(directory)
        
        with open(gitkeep_path, 'a'):
            pass
```
