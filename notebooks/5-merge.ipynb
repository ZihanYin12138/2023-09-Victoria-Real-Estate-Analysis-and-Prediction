# Step 5 - merge `domain_data` & 4 external datasets

```python
from shapely.geometry import Point
import geopandas as gpd
import pandas as pd
import ast
```

## 1. Add 'sa2_code' & 'sa2_name' to `domain_data`, using the shape file `VIC_SA2`

### 1.1 Add 2 features 'sa2_code' & 'sa2_name' to `domain_data`

import `domain_data` & the shape file `VIC_SA2`

```python
domain_data = pd.read_csv('../data/curated/domain_data/domain_data.csv')
VIC_SA2 = gpd.read_file('../data/landing/external_data/shapefile/SA2_2021_AUST_SHP_GDA2020/SA2_2021_AUST_GDA2020.shp')
```

overview of 2 datasets

```python
domain_data
```

```python
VIC_SA2
```

simple preprocessing for `VIC_SA2`

```python
VIC_SA2 = gpd.GeoDataFrame(VIC_SA2)

# only keep the data for the Victoria
VIC_SA2 = VIC_SA2[VIC_SA2['STE_NAME21'] == 'Victoria']
# create a JSON
VIC_SA2 = VIC_SA2[['SA2_CODE21', 'SA2_NAME21', 'geometry']]

VIC_SA2.reset_index(drop=True, inplace=True)
```

```python
VIC_SA2
```

rename 2 features for readability

```python
VIC_SA2.rename(columns={'SA2_CODE21': 'sa2_code'}, inplace=True)
VIC_SA2.rename(columns={'SA2_NAME21': 'sa2_name'}, inplace=True)
```

```python
VIC_SA2
```

change the data type of 'sa2_code' from string type to int type

```python
print(VIC_SA2['sa2_code'].apply(type))
```

```python
VIC_SA2['sa2_code'] = VIC_SA2['sa2_code'].astype(int)
print(VIC_SA2['sa2_code'].apply(type))
```

define a function which can convert 'coordinates' to point geometry

```python
def convert_to_point(coord_str):
    # covert string datatype to a real python list
    coord_list = ast.literal_eval(coord_str)
    
    # create and return point object
    return Point(coord_list[1], coord_list[0])
```

apply the function to the feature 'coordinates'

```python
domain_data['property_geometry'] = domain_data['coordinates'].apply(convert_to_point)
domain_data_gdf = gpd.GeoDataFrame(domain_data, geometry='property_geometry')
domain_data_gdf.head()
```

check if the 2 geometry objects in `domain_data_gdf` & `VIC_SA2` have the same datatype

```python
domain_data_gdf.set_crs(VIC_SA2.crs, inplace=True)
```

Add features about sa2 to `domain_data` based on the relationship of the property coordinates to the sa2 boundary

```python
domain_data = gpd.sjoin(domain_data_gdf, VIC_SA2, how="left", op="within")
```

```python
domain_data.head()
```

add the feature 'sa2_geometry' to `domain_data`

```python
domain_data = pd.merge(domain_data, VIC_SA2[['sa2_code', 'geometry']], on='sa2_code', how='left')

# rename the sa2 boundary geometry in `VIC_SA2` & `domain_data`
VIC_SA2.rename(columns={'geometry': 'sa2_geometry'}, inplace=True)
domain_data.rename(columns={'geometry': 'sa2_geometry'}, inplace=True)

# delete unnecessary index column
domain_data.drop(columns=['index_right'], inplace=True)
```

```python
domain_data.head()
```

## 3. Create the dataset `sa2_info`, using the 4 external datasets

```python
VIC_SA2
```

### 3.1 Create `sa2_info` by merging `VIC_SA2` & `avg_personal_income`

import the external dataset `personal_income`

```python
avg_personal_income = pd.read_csv('../data/curated/external_data/personal_income.csv')
```

overview of `avg_personal_income`

```python
avg_personal_income
```

merge to create `sa2_info`

```python
sa2_info = VIC_SA2 \
    .merge(
        avg_personal_income[['SA2', '2021-22', 'average_annual_income_growth_rate(%)']], 
        left_on = 'sa2_code', 
        right_on = 'SA2', 
        how = 'left'
    ).drop('SA2', axis=1)

sa2_info.rename(columns={'2021-22': 'personal_income'}, inplace=True)
sa2_info.rename(columns={'average_annual_income_growth_rate(%)': 'avg_income_growth_rate(%)'}, inplace=True)
```

```python
sa2_info
```

### 3.2 Merge `sa2_info` & `population`

import the external dataset `predicated_pop`

```python
population = pd.read_csv('../data/curated/external_data/predicated_pop.csv')
```

overview of `population`

```python
population
```

merge `sa2_info` & `population`

```python
sa2_info = sa2_info \
    .merge(population[['SA2_NAME', '2021', 'avg_growth_rates(%)']], left_on='sa2_name', right_on='SA2_NAME', how = 'left') \
    .drop('SA2_NAME', axis=1)

sa2_info.rename(columns={'2021': '2021_population'}, inplace=True)
sa2_info.rename(columns={'avg_growth_rates(%)': 'avg_pop_growth_rates(%)'}, inplace=True)
```

```python
sa2_info
```

### 3.3 Merge `sa2_info` & `sa2_crime_rate`

#### 2.3.1 Convert `postcode_crime_count` to `sa2_crime_count`

import the external dataset `Crime`, this contains crime count based on postcodes

```python
postcode_crime_cont = pd.read_csv('../data/curated/external_data/Crime.csv')
```

overview of `postcode_crime_cont` & `postcode_sa2_relationship`

```python
postcode_crime_cont
```

```python
postcode_sa2_relationship
```

add the 'crime_count' based on postcode, to each postcode in `postcode_sa2_relationship`, then call it `sa2_crime_cont`

Assumption: we assume the several sa2's under a postcode have the same crime rate, I will mention it again later.

```python
sa2_crime_cont = postcode_sa2_relationship.merge(postcode_crime_cont, on="postcode", how="left")
sa2_crime_cont.rename(columns={'crime_count': 'postcode_crime_count'}, inplace=True)
```

```python
sa2_crime_cont
```

#### 2.3.2 Convert `sa2_crime_cont` to `sa2_crime_rate`

overview of `population`

```python
population
```

add the 'population' based on sa2, to each sa2 in `sa2_crime_cont`

```python
sa2_crime_cont = sa2_crime_cont \
    .merge(population[['SA2_NAME', '2021']], left_on='sa2_name', right_on='SA2_NAME', how = 'left') \
    .drop('SA2_NAME', axis=1)

sa2_crime_cont.rename(columns={'2021': '2021_population'}, inplace=True)
```

```python
sa2_crime_cont
```

group by 'postcode' to calculate the population under each postcode

```python
postcode_pop = sa2_crime_cont.groupby('postcode', sort=False)['2021_population'].sum()
postcode_pop = postcode_pop.reset_index()
postcode_pop.rename(columns={'2021_population': 'postcode_population'}, inplace=True)
```

```python
postcode_pop
```

add 'postcode_population' to `sa2_crime_cont`, then call this df `sa2_crime_rate`

```python
sa2_crime_rate = sa2_crime_cont[['postcode', 'sa2_code', 'sa2_name', 'postcode_crime_count']].merge(postcode_pop, on='postcode', how = 'left')
sa2_crime_rate
```

calculate sa2 crime rate by using 'postcode_crime_count' & 'postcode_population'

Assumption: we assume the several sa2's under a postcode have the same crime rate.

```python
sa2_crime_rate['crime_rate(%)'] = sa2_crime_rate['postcode_crime_count'] / sa2_crime_rate['postcode_population'] * 100
```

```python
sa2_crime_rate
```

then we found that some postcodes have the same sa2

```python
unique_count = sa2_crime_rate['sa2_name'].nunique()
print(unique_count)
```

For a recurring 'sa2_name', average the crime rate for it.

```python
sa2_crime_rate = sa2_crime_rate.groupby(['sa2_code', 'sa2_name'], sort=False)['crime_rate(%)'].mean().reset_index()
```

```python
sa2_crime_rate
```

#### 2.3.3 Merge `sa2_info` & `sa2_crime_rate`

```python
sa2_info
```

```python
sa2_info = sa2_info.merge(sa2_crime_rate[['sa2_code', 'crime_rate(%)']], on='sa2_code', how = 'left')
```

```python
sa2_info
```

### 3.4 Store `sa2_info` to `data/curated/external_data/`

```python
output_directory = '../../real-estate-industry-project-open-source-industry-project-22/data/curated/external_data/sa2_info.csv'
sa2_info.to_csv(output_directory, index=False)
```

## 4. Merge `domain_data` & `sa2_info` to create `merged_data`

### 4.1 Create `merged_data`

```python
domain_data.head()
```

```python
selected_feature_list = [
    'sa2_name', 'personal_income', 'avg_income_growth_rate(%)', '2021_population', 'avg_pop_growth_rates(%)', 'crime_rate(%)'
]

merged_data = domain_data \
    .merge(sa2_info[selected_feature_list], on='sa2_name', how = 'left')
```

```python
merged_data
```

### 4.2 Simple preprocessing for `merged_data`

drop the feature 'desc' from `merged_data`, because we thought this feature has no useful information to extract

```python
merged_data.drop(columns=['desc'], inplace=True)
```

calculate victorial average 'personal_income' & 'income_growth_rate' to fill NaN in `merged_data`

```python
sa2_info.head()
```

```python
vic_avg_personal_income = sa2_info['personal_income'].mean()
merged_data['personal_income'].fillna(vic_avg_personal_income, inplace=True)
```

```python
vic_avg_income_growth_rate = sa2_info['avg_income_growth_rate(%)'].mean()
merged_data['avg_income_growth_rate(%)'].fillna(vic_avg_income_growth_rate, inplace=True)
```

fill NaN in the feature 'crime_rate(%)' with 0

```python
merged_data['crime_rate(%)'].fillna(0, inplace=True)
```

```python
merged_data
```

### 4.3 Store `merged_data` to `data/curated/merged_data/`

```python
output_directory = '../../real-estate-industry-project-open-source-industry-project-22/data/curated/merged_data/merged_data.csv'
merged_data.to_csv(output_directory, index=False)
```
